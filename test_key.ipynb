```python
from litellm import completion
from dotenv import load_dotenv
import os

load_dotenv()

GROQ_KEY = os.getenv("GROQ_API_KEY")
GLM_KEY = os.getenv("GLM_API_KEY")   # your existing Z.AI key, used as-is

models_to_test = {
    "Groq (gpt-oss-120b)":        ("groq/openai/gpt-oss-120b",       GROQ_KEY),
    "Kimi K2 (open-weight, via Groq)": ("groq/moonshotai/kimi-k2-instruct-0905", GROQ_KEY),
    "GLM-4.5-Flash (open, free tier)": ("zai/glm-4.5-flash",          GLM_KEY),
}

for name, (model_id, key) in models_to_test.items():
    print(f"--- Testing {name} ({model_id}) ---")
    try:
        response = completion(
            model=model_id,
            messages=[{"role": "user", "content": "Explain RAG in one sentence."}],
            api_key=key,
        )
        print("✅ Success:", response.choices[0].message.content)
    except Exception as e:
        print("❌ Failed:", str(e))
    print()
```
